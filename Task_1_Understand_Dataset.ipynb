## 6. Missing Values (incl. hidden blanks & duplicates)

```python
print("--- Null counts per column ---")
print(df.isnull().sum())
print("\nTotal nulls:", int(df.isnull().sum().sum()))

# Some CSVs store missing as blank / whitespace strings (TotalCharges is a known case)
print("\n--- Blank / whitespace-only strings per object column ---")
blank_counts = {
    c: int((df[c].astype(str).str.strip() == "").sum())
    for c in df.select_dtypes(include=["object", "string"]).columns
}
print(blank_counts)

print("\nDuplicate rows:", int(df.duplicated().sum()))
print("\nChurn distribution:")
print(df["Churn"].value_counts())
print((df["Churn"].value_counts(normalize=True) * 100).round(2))
```

## 7. Findings / Conclusion

- **Size:** 7,043 customers × 21 columns (customerID + 19 features + target `Churn`).
- **Target:** `Churn` = Yes (1,869 / 26.5%) vs No (5,174 / 73.5%) — imbalanced but usable.
- **Types:** 3 numeric (`SeniorCitizen`, `tenure`, `MonthlyCharges`); rest are object/string. `TotalCharges` is object instead of numeric — must be coerced to numeric in cleaning.
- **Missing values:** `isnull()` shows 0, but **11 blank strings in `TotalCharges`** (customers with `tenure = 0`) — hidden missing values to fix with `pd.to_numeric(errors='coerce')`.
- **Duplicates:** 0 duplicate rows; `customerID` is unique (7,043 unique).
